# Introduction to Supervised Learning

Machines learn how to solve a given problem by being provided with examples of the problem. **Supervised learning** is a specific type of machine learning, where an algorithm learns to make predictions or classify data based on labeled examples. In other words, the machine is guided by a set of input-output pairs, where the desired output (called the **label or target**) is known for each input.

<img src="materials/images/machine-learning.png"/>

For example, we provide the machine with several examples of chest X-rays, some representing normal, healthy patients and others indicating pneumonia. Then, we label the X-rays respectively.

The algorithm will learn the relationship between the examples and the labels. It will learn which collection of features within an X-ray image suggests that it is normal versus pneumonia. So the labels, in effect, guide or supervise the machine's learning.

---

<img src="materials/images/supervised-learning.png"/>

 The algorithm analyzes the provided examples, and learns the underlying patterns or relationships. Then, it builds a model that can predict the correct labels for new, unseen inputs.

<div class="alert alert-block alert-success">
<b>Note:</b> With the guidance of labeled examples, supervised learning can achieve high accuracy and make reliable predictions. </b>
</div>

---

## Structured data


Images, like the example X-rays above, are known as **unstructured** data.

**Structured** data, like that neatly organized into rows and columns, is handled a little differently when doing supervised learning. The data set below, for predicting heart disease, is an example of structured data.

<img src="materials/images/structured-data_1.png"/>

The "heart disease" column above is known as the **target variable** - it is what we are trying to predict. The remaining columns are considered the data set's **features**. The machine learning algorithm's objective is to learn to predict the target (i.e., whether a patient has heart disease), given the patient's features.

Since the target, or label, is provided to the machine, along with the various features, this is considered supervised learning. The algorithm learns to recognize patterns and associations between the features and their corresponding labels, enabling it to make predictions on new, unseen data.

Again, the machine is guided by a set of input-output (features-target) pairs. However, with structured data, we typically assign the features to the variable "X" and the target to the variable "y".

The machine learns to predict **"y"** (`i.e., whether the patient has heart disease`), given **"X"** (`i.e., the patient's age, sex, type of chest pain, blood pressure, cholesterol, maximum heart rate, and EKG - ST segment depression`).

<img src="materials/images/structured-data_2.png"/>

We then provide a machine learning algorithm with the variables X and y so that it can perform supervised learning.

# Supervised Machine Learning Project

## Data: Heart Disease

```python
import pandas as pd

df = pd.read_csv("data/heart-disease.csv")

# Transform "male" to 0 and "female" to 1.
df["sex"] = df["sex"].replace({"male":0, "female":1})
```

### Preview the heart disease data

```python
# Preview a random sample of 5 rows.
# The prediction is whether the patient has heart disease (1) or not (0).
df.sample(5)
```

---

## Separate the data into X (features) and y (target).

```python
# Separates the features into X
X = df.iloc[:,:-1].values

# Separates the target (labels) into y
y = df.iloc[:,-1].values
```

## Split the data into separate train and test sets.
We are using a 70% (training set) and 30% (testing set) split.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=1, test_size=0.3)
```

### Preview the features for the first 5 rows of the training set.

```python
X_train[:5]
```

### Preview the target (labels) for the first 5 rows of the training set.

```python
y_train[:5]
```

## Standardize the data

```python
# Standardizing the data ensures that each of the features falls within a similar range for comparison.

from sklearn.preprocessing import StandardScaler

stdsc = StandardScaler()

X_train_std = stdsc.fit_transform(X_train)
X_test_std = stdsc.transform(X_test)
```

<div class="alert alert-block alert-success">
<b>Note:</b> We standardize data so that each feature is on the same scale (within a similar range of numbers) enabling the machine to better compare and better understand the relationship among the various features. For example, the chest_pain feature ranges from 0-3 whereas max_hr ranges from around 100-200. This difference makes it difficult for the machine to compare the two features.  Once we scale the features to be between, typically, 3 and -3, the machine can better make sense of each of the features and how they relate to each other. </b>
</div>

---

## Select an algorithm to perform supervised machine learning and provide the algorithm with X and y -- the features-target pairs.
We're using the **Logistic Regression** algorithm to learn to make predictions from our examples and their labels.

```python
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression()

log_reg.fit(X_train_std, y_train)
log_reg.score(X_test_std, y_test)
```

`logreg.fit()` tells the algorithm to learn from the provided training set (features-label pairs).
The trained model then uses `logreg.score()` to evaluate its performance on the "unseen" test set.

The above trained model is approximately 75% accurate in predicting whether patients have heart disease, given their provided features.

<div class="alert alert-block alert-success">
<b>Note:</b> Logistic Regression is a machine learning algorithm that is best used when the prediction is binary (e.g., heart disease, no heart disease) as it learns probabilities. Any probabilities 50% or greater will be predicted to be in the positive class ("heart disease"), and probabilities less than 50% will be predicted to be in the negative class ("no heart disease"). </b>
</div>

# Summary: Supervised Learning
- **Strengths:** Supervised learning is highly effective when labeled data is available, as it directly learns from explicit input-output pairs. It excels in tasks where classifications are required. With the guidance of labeled examples, supervised learning can achieve high accuracy and make reliable predictions.
- **Applications:** Supervised learning is commonly used in applications such as image recognition, sentiment analysis, and other applications where labeled training data is abundant.

# 🧬 Applications with Other Data Types

### 1. Genomics: Predicting Disease Risk Based on Genomic Variants

**Scenario**: Certain genetic variants (e.g., single nucleotide polymorphisms, or SNPs) may increase the risk of developing a specific disease, such as heart disease. If you want to find out the risk of heart disease based on the genetic variants, this is what you could do:

1. **Data Collection**:

DNA samples are taken from a group of individuals who have heart disease (i.e., experimental group), and another group who do not (i.e., control group). Using genome-wide association studies (GWAS) or whole-genome sequencing, researchers identify millions of SNPs across the genome for each individual.

2. **Feature Extraction**:

Each SNP can serve as a feature, where its genotype (e.g., AA, AG, GG) can have different associated risks. If there are 1 million SNPs analyzed, there would be 1 million features for each sample.


3. **Model Training**:

- The data is divided into a training set and a testing set.
- The input is the genotypes of the SNPs, and the output is the known label, `Heart Disease` or `Control`.
- A supervised learning algorithm (like logistic regression, random forests, etc.) is trained on the training set.

4. **Prediction**:

Once the model is trained, it predicts the risk of heart disease for the samples in the testing set based on their genomic variants.

5. **Evaluation**:

The model's predictions are compared with the actual disease status of the samples in the testing set to assess its accuracy.

### 2. Epigenetics: Predicting Disease Susceptibility Based on DNA Methylation Patterns

**Scenario**: Researchers believe that certain diseases, like specific types of cancers or neurological disorders, might be associated with characteristic patterns of DNA methylation. They want to know whether a person is diagnosed with cancer or neurological disorders based on their DNA methylation patterns. They will do the following:

1. **Data Collection**:

The researchers collect DNA samples from patients diagnosed with the disease of interest (e.g., a specific type of cancer) and from healthy controls.
They then determine the methylation status of thousands of specific sites (CpG sites) across the genome for each sample using techniques like bisulfite sequencing.

2. **Feature Extraction**:

Each methylation site's status (methylated or unmethylated) can be treated as a feature. Therefore, if 10,000 CpG sites are analyzed, there are 10,000 features for each sample.

3. **Model Training**:

The data is divided into a training set and a testing set. A supervised learning algorithm (e.g., Random Forests, Gradient Boosting Machines, etc.) is trained on the training set. The input is the methylation status of the CpG sites, and the output is the known label (e.g., "disease" or "healthy").

4. **Prediction**:

Once the model is trained, it is tested on the testing set. For each DNA sample in the testing set, the model predicts whether the sample comes from a diseased patient or a healthy individual based on its methylation pattern.

5. **Evaluation**:

The model's predictions are compared to the true labels to determine its accuracy and potential clinical utility.

### 3. Transcriptomics: Predicting Disease Status Using RNAseq Data

**Scenario**: A group of researchers have RNAseq data from patients with a specific disease (let's say, a type of cancer) and from healthy controls. For each RNAseq sample, they know the disease status of the patient as "cancer" or "healthy". If they want to know whether a sample is from the cancer patient group or a healthy control group, they could do the following:


1. **Data Collection**:
- The researchers collect RNAseq data from a group of cancer patients, labeled as `cancer`.
- They also collect RNAseq data from a group of healthy individuals, labeled as `healthy`.


2. **Feature Extraction**:

Each gene's expression level can be treated as a feature. If there are 20,000 genes being analyzed, there are 20,000 features for each sample.


3. **Model Training**:
- The data is divided into a training set and a testing set.
- A supervised learning algorithm is trained on the training set. Examples of such algorithm are Support Vector Machines, Decision Trees, Neural Networks. The input is the gene expression levels, and the output is the known label, `cancer` or `healthy`.

4. **Prediction**:

Once the model is trained, it is tested on the testing set. For each RNAseq profile in the testing set, the model predicts whether the sample is from a cancer patient or a healthy individual.

5. **Evaluation**:

The predictions are compared to the true labels to assess the accuracy of the model.

### 4. Proteomics Example: Predicting Treatment Response in Cancer Patients Based on Tumor Proteomic Profiles

**Scenario**: Some cancer patients respond well to a particular treatment, while others do not. Researchers suspect that the protein expression patterns in tumor samples might provide clues about which patients are more likely to benefit from the treatment. In order to understand which group of cancer patients could benefit more from the treatment, they could do the following:

1. **Data Collection**:

Tumor protein samples are collected from patients before they undergo a specific cancer treatment.
These samples are then analyzed using techniques like mass spectrometry to quantify the abundance of thousands of proteins.

2. **Feature Extraction**:

The abundance or presence of specific proteins, or even protein modifications, can serve as features. If there are 5,000 proteins quantified, there would be 5,000 features for each sample.

3. **Model Training**:

The patients are grouped based on their response to treatment: "Responder" and "Non-responder."
The protein data is divided into a training set and a testing set.
A supervised learning algorithm (e.g., Support Vector Machines, Neural Networks, etc.) is trained on the training set. The input is the protein abundance levels, and the output is the known label ("Responder" or "Non-responder").

4. **Prediction**:

Once trained, the model predicts treatment response for the samples in the testing set based on their proteomic profiles.

5. **Evaluation**:

The model's predictions are compared to the actual treatment outcomes to assess its accuracy.
