**Large language models (LLMs)** are a type of artificial intelligence designed to understand and generate natural human language. These models are built using neural networks and deep learning techniques. They are **pre-trained** to be used for a variety of language tasks.

## Large typically refers to two different aspects of the LLM:

### Dataset size

These models are trained on an enormous amount of text. Typically, they are trained on billions of examples of language usage from throughout the Internet, learning grammar, vocabularies and world knowledges.

### Model size

The size or parameter count of these models is also enormous. Parameters are the weights or the knowledge that the machine learned from the training data. Parameter count often defines the skill or capability of a model to understand and generate text. This large parameter count allows the model to capture and learn complex patterns in language data.

<div class="alert alert-block alert-info">
<b>Note:</b> GPT-3 is the large language model ChatGPT is based on. It has more than 175 billion machine learning parameters. GPT-4 (a.k.a. ChatGPT Plus) has been estimated to have 1.76 trillion parameters, ten times larger than GPT-3.
</div>

One of the great benefits of using large language models is, given the enormous amount of data that they have been trained on, a single model can be used for different tasks including text classification, question answering, language translation, document summarization, and more.

## Self-attention

Another extremely important aspect of the transformer is what's known as its self-attention mechanism.

This is the magic part of the transformer. It allows the model to weigh the importance of each word in the sentence in relation to every other word.

Think of it as the model paying attention to different parts of the sentence based on their relevance to understanding the whole sentence. This is much how humans process language.

For example, if we read the sentence, "A dog ate the food because it was hungry", what is "it" referring to: the "dog" or "food"? Clearly **`"it"`** is referring to **`"dog"`**.

That might seem silly to you, but that task is challenging for a machine.

<img src="materials/images/self-attention-1.png"/>

---

Let's evaluate another sentence.

"A dog fed the puppy because it was hungry." Now **`"it"`** is referring to **`"puppy`**, but how do we, as humans, do that!? The transformer's self-attention mechanism enables it to perform many language tasks that humans take for granted.

<img src="materials/images/self-attention-2.png"/>

---

To better understand self-attention, observe the illustrated example below. The self-attention mechanism relates each word in a sentence to every other word to better understand the word.

In this case, it is evaluating the meaning of the word "it". Self-attention determines the word in the sentence with the largest attention value or strongest connection to the word "it". In this case, it learns that the word “it” is referring to the word "dog".

<img src="materials/images/self-attention-3.png"/>

**Self-attention** is a powerful tool for understanding the relationships and dependencies within language. It allows a model to focus on different parts of input data, considering how each part relates to all the other parts.

It helps the model capture contextual information and dependencies, making it particularly useful for understanding language, where the meaning of a word often depends on the words around it in a sentence.

Self-attention has played a pivotal role in the success of the transformer for natural language tasks. Virtually all LLMs are transformers.

# Self-supervised learning

LLMs are typically trained using a self-supervised learning approach. Self-supervised learning is a type of machine learning, where the model learns from the data itself, without relying on explicit labels or annotations provided by humans.

Instead, the model generates its own training labels from the input data and those labels can be modified as the model gets more exposure to varying examples and aspects of language.

One common self-supervised task for training LLMs is called **"Masked Language Modeling"**. In this task, a large corpus of text is used for training. During training, some of the words in the input sentences are randomly removed or "masked". The model's goal is to predict what the masked words should be based on the context of the surrounding words.

For example, given the sentence below with the masked words, the language model would probably predict something like, "The **`cheetah`** chases the **`antelope`** in the savanna." Or perhaps, "The **`lion`** chases the **`wildebeest`** in the savanna."

The model uses the probability of a given word being in that location based upon the patterns that it has been exposed to.

<img src="materials/images/masking-1.png"/>

<img src="materials/images/masking-2.png"/>

Given the above masked example, the model probably wouldn't predict, "The **`cat`** chases the **`mouse`** in the savanna."

It probably hasn't seen that pattern much, if at all. It would more likely predict, "The cat chases the mouse in the kitchen." That is a more likely pattern used in language.

Thus, from being exposed to billions of examples of how language is used, the model simply learns to predict the next word given some learned probabilities.

Self-supervised learning has been a breakthrough in training LLMs. It allows these models to acquire a broad understanding of language by independently processing, and learning from vast amounts of text from the internet.

# Prompting

When using LLMs to solve a language task, you don’t need to be an expert. And you don't need to train your own model. All you need is to think about prompt design.

**Prompting** is a crucial concept when working with Large Language Models. It is the process of instructing the model or providing context to the model such that the model generates useful output.

For example, you may provide the model with some text, and then request that the model translate or summarize the text. Alternatively, you may desire that the model simply complete some partial text. By crafting a well-structured prompt, you can customize the model's behavior to perform a specific task.

# Fine-tuning

While LLMs, as foundational language models, are versatile, their performance can vary depending on the specific task. Fine-tuning and adapting these models to particular tasks can significantly improve their performance, and make them more suitable for specialized purposes.

Fine-tuning a language model like GPT involves further training the model on a specific dataset that is tailored to a particular task.

The term "fine-tuning" is used because the model has already undergone a pre-training phase on a massive, general-purpose dataset. Fine-tuning customizes the model's behavior for a specific purpose or domain.

<div class="alert alert-block alert-info">
<b>Note:</b> Fine-tuning a foundational LLM updates its parameters/weights (knowledge) to account for the characteristics of the domain data and the specialized task the model is being trained to perform.
</div>



---

As an analogy of fine-tuning, a college student typically has a solid foundational understanding in language, in general. However, the student probably doesn't have a dependable understanding of how that language is used in a narrow and specialized context, such as in law and medicine.

The student would need to be exposed to specialized training (e.g., master's or doctoral program, law school, medical school) in order to have fine-tuned understanding and use of the language in a specific domain.

The specialized training assumes and builds upon the student's foundational understanding of language.

Following that training, the student can then understand and effectively communicate in ways that the language is used within the narrow and specialized domain.

Similarly, LLMs build upon their foundational understanding of language while being fine-tuned to the use of language for a special purpose, or within a specific domain.

# Prompt Engineering

The primary skill, when interacting with LLMs, is **prompt engineering**. Prompt engineering is an essential part of interacting with LLMs. Remember, your prompt may require a little refinement in order for the model to generate the desired response.


LLMs, like the  GPT series of models, are trained to generate human-like text based on the input they receive. Prompt engineering involves crafting input text in a way that guides the model to produce the desired output or answer.

## Tips for writing prompts:

### 1. Write clear instructions

**Instead of this:** "What is epigenetics?"

**Or this:** "Explain epigenetics, and keep it short."

**Do this:** "Provide a simple explanation of epigenetics in 2-3 sentences."

### 2. Clarify the audience to the model

**Instead of this:** "Summarize genomics."

**Do this:** "Briefly explain genomics to me. Assume that I have no background in molecular biology."

### 3. Rewrite the prompt to improve the model's response
Try using different synonyms. The right synonym can unlock the model's ability.

**Version 1:** "Provide a **basic** explanation of a volcano plot."

**Version 2:** "Provide a **simple** explanation of a volcano plot."

**Version 3:** "Provide an **easy-to-understand** explanation of a volcano plot."

### Predict the regulatory effect (i.e., activation, inhibition, no effect) of a given gene on another gene based on their DNA sequences.

**1. Data Collection**:

- Gather DNA sequences of genes involved in the cell cycle.
- Gather experimental data about which genes regulate others.

**2. Define Variables**:

For simplicity, let's consider three genes: Gene A, Gene B, and Gene C.

- Sequence S<sub>A</sub>: DNA sequence of Gene A

- Sequence S<sub>B</sub>: DNA sequence of Gene B

- Sequence S<sub>C</sub>: DNA sequence of Gene C

**3. Transformer-based Model for Sequence Analysis**:

- Input the DNA sequences into a transformer model. Each base (A, T, C, G) can be treated somewhat analogously to a word in NLP.

- The self-attention mechanism will allow the model to weigh the importance of different parts of the sequence in relation to other parts, identifying potential regulatory regions or motifs.

**4. Predict Regulatory Relationships**:
Using the transformer's output representations of the sequences, predict the relationships:

- Gene A → Gene B (Does Gene A activate, inhibit, or have no effect on Gene B?)

- Gene B → Gene C

The prediction can be made using the last layer's output of the transformer, followed by a classification head.

**5. Model Training & Evaluation**:

- Using the gathered gene sequences and their known regulatory relationships, train the transformer model to predict these relationships based on DNA sequences.
Split the data into training, validation, and test sets to evaluate the model's performance.

**6. Application**:

Once trained, this transformer-based model can predict the potential regulatory effect between any two genes based on their DNA sequences. This approach can be expanded to analyze larger genomic regions or even entire genomes.

This example combines genomics with deep learning, demonstrating how advanced NLP techniques can be applied to biological sequences. In reality, while transformers can be used for genomics, such applications would need careful data preparation, model architecture adjustments, and rigorous validation against experimental data.
