# Introduction to Unsupervised Learning

## Structured Data

Data neatly organized into rows and columns is considered **structured** data. The sample data set below, for predicting heart disease, is an example of structured data. The "heart disease" column on the right is known as the **target variable** - it is what we are trying to predict. The remaining columns are considered the data set's **features**.

<img src="materials/images/structured-data_1.png"/>

For supervised machine learning, we typically assign the features to the variable **"X"** and the target to the variable **"y"**. Therefore, the machine uses "X" (i.e., `the patient's age, sex, type of chest pain, blood pressure, cholesterol, maximum heart rate, and EKG - ST segment depression`) while being guided by "y" (the true classification) in order to learn to predict a future patient's outcome given the patient's features.

<img src="materials/images/structured-data_2.png"/>

## Unlabeled Data

However, when a data set lacks the labels that are typically used to guide the machine's learning, this is considered **unlabeled data**.

With unlabeled data, there is no "y", only "X". The machine learning algorithm must learn to make predictions strictly from information provided by the features, without any guidance or supervision from labels.

# Unsupervised Learning

**Unsupervised learning** is a type of machine learning, where an algorithm learns patterns and structures in data without any explicit guidance or labeled examples.

For example, in medicine and healthcare, unsupervised learning can be very useful for analyzing and segmenting disease in large data sets into subgroups in healthcare systems. It can also be used in heterogeneous patient populations to identify more homogeneous subgroups around the preferences and needs of rehabilitating patients, enabling more customized care.

Further, in radiology, vast numbers of medical images such as X-rays or MRIs are generated daily. Through unsupervised learning, these images can be processed to identify patterns or anomalies, and segmented into clusters based on common visual characteristics. These clusters could represent different stages or subtypes of a disease, thus aiding doctors in diagnosis and treatment planning.

Unsupervised learning aims to discover inherent structures, groupings, or similarities within the data without any supervision or prior knowledge about the correct classifications.

In other words, with unlabeled data, the machine learning algorithm must perform unsupervised learning to make its predictions.

## Clustering

A common technique used in unsupervised learning is **clustering**. The algorithm analyzes the characteristics and relationships among the examples to form clusters or categories based on their similarities.

Examples from the same class typically will have more similar features, and will tend to cluster together and be in closer proximity to each other. These learned similarities or clusters are what dictate the algorithm's class predictions.

To perform clustering, each data point is projected out into *n*-dimensional space (*n* is the number of columns/ features in the data set), as illustrated below.

<img src="materials/images/cluster-1.png"/>

In general, data points will be assigned to the same cluster as the data points that it is closest to. In other words, data points that share the most similarity of features will be nearest to each other and thus be assigned to the same cluster.

For example, with K-means clustering, points known as **centroids** are randomly placed out into this n-dimensional space. Each data point is initially assigned to the centroid that it is closest to. The centroid then moves to the average or "mean" location of the data points belonging to its cluster, hence the name centroid. The data points will next assess which centroid they are now closets to and, if necessary, change clusters. This process repeats until no data point changes its cluster. Each data point should now belong to the cluster with the most similarity to its features.

Each cluster's centroid is shown below. The centroid is "learned" from the data. The cluster that a data point belongs to will dictate the label or classification that will be predicted for it.

<img src="materials/images/cluster-2.png"/>

## Artificially Generated Data

Here, we will artificially generate two columns (features) of data consisting of three clusters. Two columns will enable the data set to easily be visualized.

```python
import pandas as pd
import numpy as np
from sklearn.datasets import make_blobs

X, y = make_blobs(n_samples=150, n_features=2,  centers=3,   cluster_std=0.5,   shuffle=True, random_state=0)
df = pd.DataFrame(data=X, columns=["column_1", "column_2"])

# Preview the first 10 rows of the data set
df.head(10)
```

The above two-dimensional data set does not have class labels, so a machine learning algorithm would need to perform unsupervised learning in order to classify or predict a label for each row.

## Visualize the Two-Dimensional Dataset

Below, each data point from the two-dimensional data set is visualized in its respective location on the `Feature_1-axis` and `Feature_2-axis`.

You can see how the data points form clusters based upon similarity of features. Each data point would be predicted to belong to the class represented by the cluster that it is closest to.

```python
import matplotlib.pyplot as plt
plt.scatter(X[:, 0], X[:, 1], c='white', marker='o', edgecolor='black', s=50)
plt.xlabel('Feature_1')
plt.ylabel('Feature_2')
plt.grid()
plt.tight_layout()
```

## Perform K-Means Clustering

A common unsupervised machine learning algorithm is *K-means* clustering. Here, we employ K-means on the above generated data set, and visualize the learned clusters.

```python
from sklearn.cluster import KMeans
km = KMeans(n_clusters=3, n_init="auto", random_state=0)
y_km = km.fit_predict(X)

# Visualize learned clusters
plt.scatter(X[y_km == 0, 0], X[y_km == 0, 1], s=50, c='limegreen', marker='s',
            edgecolor='black', label='Cluster 1')
plt.scatter(X[y_km == 1, 0], X[y_km == 1, 1], s=50, c='orange', marker='o',
            edgecolor='black', label='Cluster 2')
plt.scatter(X[y_km == 2, 0], X[y_km == 2, 1], s=50, c='dodgerblue', marker='v',
            edgecolor='black', label='Cluster 3')
plt.scatter(km.cluster_centers_[:, 0], km.cluster_centers_[:, 1], s=250, marker='*', c='red',
            edgecolor='black',label='Centroids')

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend(scatterpoints=1)
plt.grid()
plt.tight_layout()
```

# Unsupervised Machine Learning Project

## Data: Heart Disease

We will use a heart disease data set to learn to predict whether a patient has heart disease or not. Since there is no `heart_disease` target variable (i.e., labels), we will need to use unsupervised learning.

```python
import pandas as pd

df = pd.read_csv("data/heart-disease.csv")
# Transform "male" to 0 and "female" to 1.
df["sex"] = df["sex"].replace({"male":0, "female":1})

# Preview the first 10 rows of the heart disease data set
df.iloc[:, :-1].head(10)
```

We will begin with using just `Age` and `Max Heart Rate` to create a two-dimensional data set so that it can be visualized.

Below, we selected two columns (`age` and `max_hr`) from the heart disease data set, and placed them into the variable X.

```python
X = df[['age', 'max_hr']].values

# Preview the first 10 rows of the two-dimensional data set (i.e., "age", "max_hr")
X[:10]
```

### Visualize the 2-Dimensional Dataset

Below, we create a scatter-plot of the data points prior to being clustered by an unsupervised learning algorithm.

```python
import matplotlib.pyplot as plt
plt.scatter(X[:, 0], X[:, 1], c='white', marker='o', edgecolor='black', s=50)
plt.xlabel('Age')
plt.ylabel('Max HR')
plt.grid()
plt.tight_layout()
```

## Perform K-Means Clustering

K-means is a common unsupervised learning algorithm that learns to cluster the data points based upon their proximity to a centroid.

With K-means clustering, we must select the number of clusters that we want our data grouped into. In this case, since we want to predict either "heart disease" or "no heart disease", we will choose two-means clustering. The "mean" location of a cluster is synonymous with "centroid".

### Standardize the Data

```python
# Standardizing the data ensures that each of the features is on a similar scale for distance comparisons.

from sklearn.preprocessing import StandardScaler
stdsc = StandardScaler()
X_std = stdsc.fit_transform(X)
```

### K-means unsupervised learning

```python
from sklearn.cluster import KMeans
km = KMeans(n_clusters=2, n_init="auto", random_state=0)
y_km = km.fit_predict(X_std)
```

### Visualize the Two Clusters That were Learned from the K-means Algorithm

Note the change in scale on the axes.

```python
plt.scatter(X_std[y_km == 0, 0], X_std[y_km == 0, 1], s=50, c='lightgreen', marker='s',
            edgecolor='black', label='Cluster 1')
plt.scatter(X_std[y_km == 1, 0], X_std[y_km == 1, 1], s=50, c='orange', marker='o',
            edgecolor='black', label='Cluster 2')

plt.scatter(km.cluster_centers_[:, 0], km.cluster_centers_[:, 1], s=250, marker='*', c='red',
            edgecolor='black',label='Centroids')

plt.xlabel('Age')
plt.ylabel('Max Heart Rate')
plt.legend(scatterpoints=1)
plt.grid()
plt.tight_layout()
```

The data points in the green cluster will be classified (predicted) as not having heart disease. Those in the orange cluster will be classified as having heart disease.

# Summary: Unsupervised Learning

- **Strengths**: Unsupervised learning is valuable when working with unlabeled or unstructured data, as it discovers hidden patterns and structures autonomously. It can reveal unknown relationships, detect anomalies, and uncover insights without the need for labeled data. Unsupervised learning is also useful for exploratory data analysis and data preprocessing.

- **Applications**: Unsupervised learning is often used in areas like patient segmentation, disease grouping, anomaly detection, and data visualization. It is particularly helpful in scenarios where labeled data is scarce or costly to obtain.

# ⌚ Application: Unsupervised Learning for Wearable (Time-Series) Data

Data that is recorded repeatedly at many points in time is considered a time series. Wearable data is an example of time-series data.

## Use case: Alerting system for respiratory illnesses and other stress events using wearable data

The ability to detect infectious diseases pre-symptomatically or asymptomatically enables early intervention, and can reduce transmission. When the body is fighting off an infection, like COVID-19 or the flu, the heart might be needing to work harder and beat faster to fight off the germs.

Wearable devices, such as smartwatches, have the ability to monitor the heart rate continuously in real time. One way to see if the body is working harder to fight the disease is by checking **overnight resting heart rate**, which tends to perform at a baseline cadence when the body is at rest and healthy. However, when the body is sick with an infection, the resting heart rate often becomes elevated.

Therefore, the use of wearable devices for continuous monitoring of heart rate can provide a clue about whether the body is dealing with an infection. If the heart rate is significantly higher than the baseline rate, it could be an indicator that the body is working to fight off an infection.

Our goal is to build a predictive model, using smartwatch-based time-series data, that detects aberrant resting heart rates (RHR) associated with the onset of early infection. Recognizing these changes is the domain of **unsupervised learning which identifies anomalies or outliers in time-series data**.

---

## 1. Wearable heart rate data
The following dataset is a sample from a study that performed real-time monitoring of RHR for detecting abnormal physiological events using a smartwatch.

```python
import pandas as pd
import numpy as np
import pyod
import matplotlib.pyplot as plt

wearable_data = pd.read_csv("data/P100300/Orig_Fitbit_HR.csv", index_col='datetime', parse_dates=True)
# Resample the data to determine the median overnight RHR for all dates
wd = wearable_data.resample('6H').median()[::4]
# Drop dates with missing readings (NaN)
wd = wd.dropna()
```

### Preview the wearable dataset
Median overnight heart rates.

```python
wd.head(10)
```

# 2. Detecting Outliers

For our wearable heart rate dataset, we do not know the outliers (labels) in advance, so detecting the outliers will be an unsupervised learning task.

### 2.1 Using K-Nearest Neighbors Algorithm

The K-Nearest Neighbors (KNN) algorithm can be an effective choice when working with time-series data in an unsupervised learning manner. For every data point, the algorithm finds the closest K-nearest neighbors and measures the average distance - we will set our algorithm to measure the 5-nearest neighbors to each heart rate. A threshold is determined and the points with the most significant distance from the time series will be considered outliers.

```python
from pyod.models.knn import KNN

# Here, we will use the mean of the distances to the 5-nearest neighbors for each data point to determine outliers.
knn = KNN(contamination=0.03, method='mean', n_neighbors=5)
knn.fit(wd)
predicted = pd.Series(knn.predict(wd), index=wd.index)
outliers = predicted[predicted == 1]
outliers = wd.loc[outliers.index]
```

### 2.2 Display the dates with outliers (anomolies)

```python
outliers
```

An alerting system could be used to alert the wearer of these unusually elevated resting heart rates (RHRs).

### 2.3 Prepare to visualize outliers

```python
def plot_outliers(outliers, data,
                 halignment = 'right',
                 valignment = 'top',
                 labels=False):

    ax = data.plot(alpha=0.6, figsize=(15,5))

    if labels:
        for i in outliers["heartrate"].items():
            plt.plot(i[0], i[1], 'v', markersize=8, markerfacecolor='crimson', markeredgecolor='k')
            plt.text(i[0], i[1]-(i[1]*0.04), f'{i[0].strftime("%m/%d")}', fontsize=8,
                         horizontalalignment=halignment,
                         verticalalignment=valignment)
    else:
        data.loc[outliers.index].plot(ax=ax, style='X', color="crimson", markersize=9)

    plt.title(f'Wearble RHR (overnight) - KNN')
    plt.xlabel('Date', fontsize=14)
    plt.ylabel('RHR', fontsize=14)
    plt.legend(['wearable RHR','outliers'])
    plt.show()
```

### 2.4 Plot the time series highlighting the outliers

```python
plot_outliers(outliers, wd)
```

### 2.5 Plot the time series highlighting the outliers with the dates labeled

```python
plot_outliers(outliers, wd, labels=True)
```

If the data was collected in real time, an alerting system could be used to notify the wearer as the anomolies, associated with the onset of early infection, were identified. </b>
</div>

Unsupervised learning is an effective means of identifying outliers or anomalous events in time-series data. In this case, it was used with a wearable device for continuous monitoring of heart rate to provide early detection of potential respiratory illnesses and other infections.

Resting heart rate is not a perfect measure because other stressors are also related to its elevation  (e.g., intense exercise, alcohol consumption, altitude change, travel, stress). However, it can be an alert to take preventative action to perform self-isolation, get tested, and be treated by medical professionals.

# 🧬 Applications with Other Data Types

### 1. Genomics: Discovering Genetic Subtypes in a Specific Disease Using Genomic Variants

**Scenario**: Researchers believe that a particular disease, let's say breast cancer, may have distinct genetic subtypes that are yet to be discovered. These subtypes might differ in their genomic makeup, and recognizing them could lead to more targeted treatments. In order to recognize these subtypes, you could use unsupervised learning algorithm to recognize them based on the SNP of a group of cancer patients. Then, you can further investigate how each subtype progress in disease or respond to treatment.

1. **Data Collection**:

DNA samples are collected from a group of breast cancer patients.
These samples are analyzed using whole-genome sequencing or SNP arrays to identify millions of SNPs across the genome for each individual.

2. **Feature Extraction**:

Each SNP or genetic variant can serve as a feature, with its genotype (e.g., AA, AG, GG) representing different states.
If 1 million SNPs are analyzed, there would be 1 million features for each sample.

3. **Pattern Recognition** (Clustering):

An unsupervised learning algorithm, like k-means clustering or hierarchical clustering, is applied to the genomic data.
The algorithm groups patients based on the similarity of their genomic variants without any prior labels.

4. **Discovery**:

The clustering results might reveal distinct groups or clusters of patients.
These clusters might represent different genetic subtypes of breast cancer.

5. **Validation and Interpretation**:

Researchers then analyze these clusters to see if they have clinical significance. For instance, patients in one cluster might have a more aggressive form of the disease or a better response to a specific treatment.

Further biological studies might be done to understand the genetic differences underlying each subtype.

### 2. Epigenetics: Uncovering Epigenetic Patterns Associated with Aging

**Scenario**: As individuals age, various epigenetic changes, such as DNA methylation patterns, may evolve across their genomes. Researchers suspect that there might be distinct patterns of methylation that could serve as "epigenetic clocks" or markers of biological aging. If you want to understand how methylation patterns are correlated with aging, this is what you could do:

1. **Data Collection**:

- DNA samples are taken from individuals across various age groups, from infants to the elderly.
- These samples are analyzed using bisulfite sequencing to measure the methylation status of millions of CpG sites across the genome.

2. **Feature Extraction**:

Each CpG site's methylation status serves as a feature, either being methylated or unmethylated.
If 2 million CpG sites are analyzed, there would be 2 million features for each sample.

3. **Pattern Recognition** (Clustering):

An unsupervised learning algorithm, like principal component analysis (PCA) followed by k-means clustering, is applied to the methylation data.
The algorithm attempts to group individuals based on the similarity of their DNA methylation patterns, without using their actual age as a label.

4. **Discovery**:

The clustering results might reveal distinct groups or clusters of individuals.
For instance, clusters might unintentionally align with specific age groups or life stages, suggesting common epigenetic patterns associated with those ages.

5. **Validation and Interpretation**:

- Researchers then examine these clusters to identify if they correlate with biological age, health status, or susceptibility to age-related diseases.
- They might find that certain methylation patterns are consistent markers of aging, offering insights into the biological processes of aging and potential targets for interventions.

### 3. Transcriptomics: Discovering Novel Subtypes of a Disease Based on Transcriptome Profiles

**Scenario**: Researchers believe that a specific disease, let's use depression as an example, might have subtypes that are not yet classified based on traditional clinical criteria. These subtypes might have distinct transcriptomic signatures. In order to recognize the subtypes of depression, you can use unsupervised learning algorithm to categorize RNA-seq data collected from a group of patients with depression. You can then investigate each group's symptoms and how they respond the treatments.

1. **Data Collection**:

- RNA samples are collected from brain tissue or peripheral blood of individuals diagnosed with depression.
- These samples are analyzed using RNA sequencing (RNAseq) to measure the expression levels of thousands of genes.

2. **Feature Extraction**:

Each gene's expression level serves as a feature.
If 25,000 genes are analyzed, there would be 25,000 features for each sample.

3. **Pattern Recognition** (Clustering):

- An unsupervised learning algorithm, like t-SNE (t-distributed stochastic neighbor embedding) followed by hierarchical clustering, is applied to the RNA-seq data.
- The algorithm attempts to group individuals based on the similarity of their gene expression profiles, without any pre-existing labels regarding the subtypes of depression.

4. **Discovery**:

- The clustering results may reveal distinct groups or clusters of individuals.
- These clusters might represent different molecular subtypes of depression, each with a unique transcriptomic signature.

5. **Validation and Interpretation**:

- Researchers then investigate these clusters to determine if they correlate with any clinical features, such as severity, treatment response, or specific symptoms.
- They might find that individuals in a particular cluster have a common symptom or respond similarly to a specific treatment, suggesting a shared biological mechanism underpinning that subtype.

### 4. Proteomics: Identifying Protein Expression Patterns in Response to Drug Treatment

**Scenario**: Scientists are studying the effect of a new drug on cells. They suspect that the drug induces changes in protein expression, but they aren't sure which proteins are affected or how. Using unsupervised learning, they aim to identify patterns of protein expression in cells treated with the drug.

1. **Data Collection**:

Cells are treated with the new drug, and protein samples are collected at various time points post-treatment.
These samples are then analyzed using mass spectrometry to quantify the abundance of thousands of proteins.

2. **Feature Extraction**:

The abundance or presence of specific proteins can serve as features. If there are 10,000 proteins quantified, there would be 10,000 features for each sample.

3. **Pattern Recognition** (Clustering):

- An unsupervised learning algorithm, like PCA (principal component analysis) followed by k-means clustering, is applied to the proteomic data.
- The algorithm aims to group samples based on the similarity of their protein expression profiles, without any predefined categories or expectations regarding the drug's effects.

4. **Discovery**:

- The clustering results might reveal distinct patterns or clusters of protein expression over time.
- For instance, certain proteins might consistently increase in abundance after drug treatment, while others might decrease.

5. **Validation and Interpretation**:

- Researchers then study these protein expression patterns to understand the drug's mechanism of action. For instance, if proteins related to apoptosis (programmed cell death) are upregulated, the drug might induce cell death.
- Further biological validation can be done to confirm these findings, such as using functional assays to test the drug's impact on apoptosis.
